```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import r2_score
from sklearn.model_selection import GridSearchCV
```

```python
df=pd.read_csv(r"C:\Users\THAJIN T\Downloads\crop_production.csv\crop_production.csv")
df
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.isnull().sum()
```

```python
df
```

```python
ob1=LabelEncoder()
ob2=LabelEncoder()
ob3=LabelEncoder()
ob4=LabelEncoder()
```

```python
df['State_Name']=ob1.fit_transform(df['State_Name'])
df['District_Name']=ob2.fit_transform(df['District_Name'])
df['Season']=ob3.fit_transform(df['Season'])
df['Crop']=ob4.fit_transform(df['Crop'])
```

```python
df
```

```python
x=df.drop(['Production'],axis=1)
x
```

```python
y=df.Production
y
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3,random_state=42)
xtrain
```

```python
ytrain
```

```python
X=np.asarray(xtrain)
Y=np.asarray(ytrain)
```

```python
model=RandomForestRegressor()
parameters={'n_estimators':[100,150,200,125,250,300]}
```

```python
forest=GridSearchCV(model,parameters,cv=5)
forest.fit(X,Y)
```

```python
best_parameters=forest.best_params_
print(best_parameters)
```

```python
highest_accuracy=forest.best_score_
print(highest_accuracy)
```

```python
final=RandomForestRegressor(n_estimators=200)
final.fit(xtrain,ytrain)
```

```python
pred=final.predict(xtest)
pred
```

```python
final.score(xtrain,ytrain)
```

```python
final.score(xtest,ytest)
```
